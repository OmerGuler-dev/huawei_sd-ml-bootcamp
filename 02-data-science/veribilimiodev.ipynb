```python
# 6. soru doldurmadan önceki boş değerler
df_temiz = df.copy()
print(df_temiz[["indirim_orani", "musteri_puani"]].isnull().sum())
```

```python
df_temiz["musteri_puani"] = df_temiz["musteri_puani"].fillna(df_temiz["musteri_puani"].median())
```

```python
#boşlukları medyan ile doldurduktan sonraki boş değerler
print(df_temiz[["indirim_orani", "musteri_puani"]].isnull().sum())
```

```python
# 7.soru çözümü
df_temiz["odeme_turu"] = df_temiz["odeme_turu"].fillna(df_temiz["odeme_turu"].mode()[0])
```

```python
df_temiz["musteri_tipi"] = df_temiz["musteri_tipi"].fillna(df_temiz["musteri_tipi"].mode()[0])
```

```python
print(df_temiz[["odeme_turu", "musteri_tipi"]].isnull().sum())
```

```python
def metin_normalize(series):

  return(
      series
      .str.strip()
      .str.lower()
      .str.replace("i̇", "i", regex=False)
      .str.replace("ı", "i", regex=False)
      .str.replace("ö", "o", regex=False)
      .str.replace("ü", "u", regex=False)
      .str.replace("ç", "c", regex=False)
      .str.replace("ş", "s", regex=False)
      .str.replace("ğ", "g", regex=False)
  )
```

```python
df["sehir"] = metin_normalize(df["sehir"])
df["sehir"].unique()
```

```python
#11 . soru kategori düzenleme
df["kategori"].unique()
```

```python
df["kategori"] = metin_normalize(df["kategori"])
df["kategori"].unique()
```

```python
kategori_duzeltme = {
    "ev&yasam" : "ev ve yasam"
}
df["kategori"] = df["kategori"].replace(kategori_duzeltme)
df["kategori"].unique()
```

```python
# 13. soru
Q1 = df["birim_fiyat"].quantile(0.25)
Q3 = df["birim_fiyat"].quantile(0.75)

IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR
```

```python
aykiri_degerler = df[
    (df["birim_fiyat"] < alt_sinir) |
    (df["birim_fiyat"] > ust_sinir)
]
aykiri_degerler
```
