
# Kayıp Veri (Missing Value)

Amaç:

    1. Eksik veri tespiti ve farklı eksik veri çözüm yöntemlerini uygulamak
    2. Eğitimde kullanılacak güncel veri setini kaydetme

Adımlar:

    1. Veri setini yükleme
    2. Eksik verileri tespit etme
    3. Eksik veri oranlarını hesaplama
    4. Eksik veri içeren satırları incele
    5. Eksik verileri silmek
    6. Sayısal eksik verileri ortalama ile doldur
    7. Sayısal eksik verileri medyan ile doldur
    8. Sayısal eksik verileri mod ile doldur
    9. Kategorik eksik verileri mod ile doldur
    10. Kategorik eksik verileri sabit bir değer ile doldurmak
    11. Eğitimde kullanılacak olan güncel veri setini oluştur
    12. Düzenlenmiş veri setini kaydet



```python
import pandas as pd
```

```python
# 1. veri setini yükle
df = pd.read_csv("e_ticaret_veri_seti.csv")
```

```python
# 2. eksik verileri tespit etme
df.isnull().sum()
```

```python
#eksik veri oranlarını hesapla
eksik_sayilari = df.isnull().sum()
eksik_oranlari = (eksik_sayilari / len(df)) * 100
eksik_oranlari
```

```python
# 4. eksik veri olan satırları inceleme
eksik_veri_olan_satirlar = df[df.isnull().any(axis = 1)]
eksik_veri_olan_satirlar.head(10)
```

```python
# 5. eksik verileri silme
df_silinen = df.dropna() # eksik veri içeren satırları veri setinden çıkardık
df_silinen.shape
```

```python
# sayısal verileri ortalama ile doldurma
df_ortalama = df.copy()
sayisal_sutunlar = df_ortalama.select_dtypes(include = ["int64", "float64"]).columns
sayisal_sutunlar
```

```python
for sutun in sayisal_sutunlar:
  df_ortalama[sutun] = df_ortalama[sutun].fillna(df_ortalama[sutun].mean())
```

```python
df_ortalama[sayisal_sutunlar].isnull().sum()
```

```python
# sayisal eksik verileri medyan ile doldurma
df_medyan = df.copy()
for sutun in sayisal_sutunlar:
  df_medyan[sutun] = df_medyan[sutun].fillna(df_medyan[sutun].median())

df_medyan[sayisal_sutunlar].isnull().sum()
```

```python
# sayısal eksik verileri mod ile doldurma

df_mod = df.copy()
for sutun in sayisal_sutunlar:
  mod_degeri = df_mod[sutun].mode()[0]
  df_mod[sutun] = df_mod[sutun].fillna(mod_degeri)
df_mod[sayisal_sutunlar].isnull().sum()
```

```python
# 9. kategorik verileri mod ile doldurma
df_mod_kategorik = df.copy()
kategorik_sutunlar = df_mod_kategorik.select_dtypes(include = ["object"]).columns
kategorik_sutunlar
```

```python
for sutun in kategorik_sutunlar:
  mod_degeri = df_mod_kategorik[sutun].mode()[0]
  df_mod_kategorik[sutun] = df_mod_kategorik[sutun].fillna(mod_degeri)
df_mod_kategorik[kategorik_sutunlar].isnull().sum()
```

```python
#10 kategorik eksik verileri sabit değer ile doldurma
df_sabit_kategorik = df.copy()

for sutun in kategorik_sutunlar:
  df_sabit_kategorik[sutun] = df_sabit_kategorik[sutun].fillna("bilinmiyor")

df_sabit_kategorik[kategorik_sutunlar].isnull().sum()
```

```python
# 11 eğitimde kullanılacak güncel veri seti
# sayısal eksik verileri medyan ile dolduracağız
#kategorik eksik veriler mod ile doldur
```

```python
df_guncel = df.copy()
```

```python
for sutun in sayisal_sutunlar:
  df_guncel[sutun] = df_guncel[sutun].fillna(df_guncel[sutun].median())
```

```python
for sutun in kategorik_sutunlar:
  mod_degeri = df_guncel[sutun].mode()[0]
  df_guncel[sutun] = df_guncel[sutun].fillna(mod_degeri)
```

```python
df_guncel.isnull().sum()
```

```python
import pandas as pd
```

```python
#1. veri setini yükleme
df = pd.read_csv("e_ticaret_veri_seti_veri_tipleri_duzenlendi.csv")
```

```python
#2. Tekrarlayan kayıtlar tespit
tekrarlayan_kayitlar = df[df.duplicated()]
tekrarlayan_kayitlar
```

```python
#3. tekrarlayan kayıtlar incele
tekrarlayan_kayitlar.head(10)
```

```python
#4. tekrarlayan kayıtları temizleme
df = df.drop_duplicates()
df.shape
```

```python
#5. tutarsız değerlerin incelenmesi
df["sehir"].unique()
#istanbul , Ankara , Gaziantep , İzmir problemli
```

```python
df["kategori"].unique()
#spor , ofis , kitap , ev yaşam , kozmetik , Elektronik problemli
```

```python
df["musteri_tipi"].unique()
#yeni , vip , mevcut problemli
```

```python
# Yardımcı fonksiyon , metinleri normalize etme
#büyük küçük harf , boşlukların kaldırılması , Türkçe karakter kaynaklı farkların düzenlenmesi
def metin_normalize(series):

  return(
      series.astype(str)
      .str.strip()
      .str.lower()
      .str.replace("i̇", "i", regex=False)
      .str.replace("ı", "i", regex=False)
      .str.replace("ö", "o", regex=False)
      .str.replace("ü", "u", regex=False)
      .str.replace("ç", "c", regex=False)
      .str.replace("ş", "s", regex=False)
      .str.replace("ğ", "g", regex=False)
  )
```

```python
# 6.şehir sütunundaki tutarsız değerleri güncelleme
df["sehir"] = metin_normalize(df["sehir"])
df["sehir"].unique()
```

```python
sehir_duzeltme_haritasi = {
    "istanbul": "İstanbul",
    "ankara": "Ankara",
    "izmir": "İzmir",
    "antalya": "Antalya",
    "konya": "Konya",
    "bursa": "Bursa",
    "adana": "Adana",
    "gaziantep": "Gaziantep",
    "mersin": "Mersin",
    "kayseri": "Kayseri",
    "eskisehir": "Eskişehir",
    "samsun": "Samsun",
    "trabzon": "Trabzon",
    "kocaeli": "Kocaeli"
}
df["sehir"] = df["sehir"].replace(sehir_duzeltme_haritasi)
df["sehir"].unique()
```

```python
#7 . kategori sütunu normalize
df["kategori"] = metin_normalize(df["kategori"])
df["kategori"].unique()
```

```python
kategori_duzeltme_haritası = {
    "ev&yasam" : "ev ve yasam"
}
df["kategori"] = df["kategori"].replace(kategori_duzeltme_haritası)
df["kategori"].unique()
```

```python
#8 ödeme türü
df["odeme_turu"] = metin_normalize(df["odeme_turu"])
df["odeme_turu"].unique()
```

```python
odeme_turu_duzeltme_haritasi = {
    "kredi karti": "Kredi Kartı",
    "kart": "Kredi Kartı",
    "banka karti": "Banka Kartı",
    "dijital cuzdan": "Dijital Cüzdan",
    "kapida odeme": "Kapıda Ödeme",
    "havale": "Havale/EFT",
    "eft": "Havale/EFT",
    "havale/eft": "Havale/EFT"
}
df["odeme_turu"] = df["odeme_turu"].replace(odeme_turu_duzeltme_haritasi)
df["odeme_turu"].unique()
```

```python
#9. müsteri tipindeki düzeltmeler
df["musteri_tipi"] = metin_normalize(df["musteri_tipi"])
df["musteri_tipi"].unique()
```

```python
musteri_tipi_duzeltme_haritasi = {
    "mevcut": "Mevcut",
    "yeni": "Yeni",
    "vip": "VIP"
}
df["musteri_tipi"] = df["musteri_tipi"].replace(musteri_tipi_duzeltme_haritasi)
df["musteri_tipi"].unique()
```

#Aykırı Değerleri İnceleme

Amaç :
  1. veri setindeki aykırı değerlerin tespit edilmesi
  2. aykırı değerleri temizleyerek güncel veri setleri oluşturulur

Adımlar :
  1. Veri setini yükle
  2. Sayısal sütunları belirle
  3. sayisal sütunları özet istatistiklerini hesapla
  4. iqr ile aykırı değer sınırlarını hesapla
  5. aykırı değer içeren kayıtları tespit etmek
  6. aykırı değer sayılarını sütun bazında incele
  7. aykırı değer içeren kayıtları veri setinden temizlemek
  8. Temzilenmiş veri setini kaydetmek
  

```python
import pandas as pd
```

```python
df = pd.read_csv("e_ticaret_veri_seti_tekrarlayan_kayitlar_duzenlendi.csv")
```

```python
#sayısal sütunlarn belirlenmesi
sayisal_sutunlar = df.select_dtypes(["int64","float64"]).columns.tolist()
sayisal_sutunlar
```

```python
#sayisal sütunların özet istatistiklerini inceleyelim
df[sayisal_sutunlar].describe()
```

```python
"""
birim fiyat
IQR = Q3(75%) - Q1(25%)
817 - 198 = 619

üst sınır = Q3(75%) + k * IQR
alt sınır = Q1(%25) + k * IQR (k  değeri genelde 1.5)
k = 1 olsun
    Üst sınır = 817 + 619 = 1436
    Alt sınır = 198 - 619 = -421
  Outlier = üst sınır ve alt sınır dışında kalan örnekler
"""
```

```python
# 4. IQR ile aykırı değer sınırlarını hesaplama
aykiri_sinirlar =  {}

for sutun in sayisal_sutunlar:

  q1 = df[sutun].quantile(0.25)
  q3 = df[sutun].quantile(0.75)

  iqr = q3 - q1
  alt_sinir = q1 - 1.5 * iqr
  ust_sinir = q3 + 1.5 * iqr

  aykiri_sinirlar[sutun] = {
      "q1" : q1,
      "q3" : q3,
      "iqr" : iqr,
      "alt_sinir" : alt_sinir,
      "ust_sinir" : ust_sinir
  }

aykiri_sinirlar
```

```python
#5. aykırı değer içeren kayıtları tespit et
aykiri_maskesi = pd.Series(False, index = df.index)

for sutun in sayisal_sutunlar:
  alt_sinir = aykiri_sinirlar[sutun]["alt_sinir"]
  ust_sinir = aykiri_sinirlar[sutun]["ust_sinir"]

  sutun_maskesi = aykiri_maskesi | (df[sutun] < alt_sinir) | (df[sutun] > ust_sinir)
  aykiri_maskesi = aykiri_maskesi | sutun_maskesi
```

```python
aykiri_kayitlar = df[aykiri_maskesi]
aykiri_kayitlar
```

```python
# 6. aykırı değer sayılarını sütun bazında incele
for sutun in sayisal_sutunlar:
  alt_sinir = aykiri_sinirlar[sutun]["alt_sinir"]
  ust_sinir = aykiri_sinirlar[sutun]["ust_sinir"]

  aykiri_sayisi = ((df[sutun] < alt_sinir) | (df[sutun] > ust_sinir)).sum()
  print(f"Sütun aykiri sayisi = {aykiri_sayisi}")
```

```python
# 7. aykiri değer içeren kayıtları temizleme
df_temiz = df[~aykiri_maskesi].copy()
df_temiz.shape
```

```python
# 8. temizlenmiş verileri kaydetme
df_temiz.to_csv("e_ticaret_veri_seti_temizlendi.csv")
```

# Kategorik Veri Analizi

Amaç :

```python
import pandas as pd
import matplotlib.pyplot as plt # görsel
import seaborn as sns # görsel
```

```python
df = pd.read_csv("e_ticaret_veri_seti_temizlendi.csv")
```

```python
# kategorik sütunlar
kategorik_sutunlar = df.select_dtypes(include = ["object","category"]).columns.tolist()
kategorik_sutunlar
```

```python
#en az görünen kategoriler
for sutun in kategorik_sutunlar:
  print(df[sutun].value_counts().tail(3))
  print(" ")
```

```python
# grafikselleştirilmesi
for sutun in detayli_analiz_sutunlari:
  if sutun in df.columns:
    plt.figure(figsize = (6,3))
    sns.countplot(data = df , x = sutun , order = df[sutun].value_counts().index)
    plt.title(f"{sutun} sütunu için count plot")
    plt.xlabel(sutun)
    plt.ylabel("Frekans")
    plt.xticks(rotation = 45)
    plt.tight_layout()
    plt.show()
```

# Pivot Tablo Analizi

Amaç:

    Veri setini iki yönli kırılımlar ile özetleyerek pivot tablo mantığını kavramak
    Pivot tablo ile karşılaştırmalı analizler ve sonuçların yorumlanması

Adımlar:

    veri setini yükle
    Kategori ve müşteri tipine göre pivot tablo analizi
    Şehir ve ödeme türüne göre pivot tablo analizi
    Kategori ve ödeme türüne göre pivot tablo analizi
    Şehir ve müşteri tipine göre pivot tablo analizi


```python
import pandas as pd
```

```python
df = pd.read_csv("e_ticaret_veri_seti_temizlendi.csv")
```

```python
#kategori ve müsteri tipine göre pivot tablo analizi
pivot_kategori = pd.pivot_table(
    df,
    index = "kategori",
    columns = "musteri_tipi",
    values = "toplam_tutar",
    aggfunc = "mean"
).round(1)
pivot_kategori
```

```python
pivot_sehir_odemeturu = pd.pivot_table(
    df,
    index = "sehir",
    columns = "odeme_turu",
    values = "toplam_tutar",
    aggfunc = "sum"
).round(1)
pivot_sehir_odemeturu
```

```python
pivot_kategori_odeme = pd.pivot_table(
    df,
    index = "kategori",
    columns = "odeme_turu",
    values = "musteri_puani",
    aggfunc = "mean"
).round(2)
pivot_kategori_odeme
```

# ZAMAN SERİSİ ANALİZİ

```python
import pandas as pd
df = pd.read_csv("e_ticaret_veri_seti_temizlendi.csv")
df.head()
```

```python
df["siparis_tarihi"] = pd.to_datetime(df["siparis_tarihi"])
```

```python
df["siparis_yili"] = df["siparis_tarihi"].dt.year
df["siparis_ayi"] = df["siparis_tarihi"].dt.month
df["siparis_gunu"] = df["siparis_tarihi"].dt.day
df["haftanin_gunu"] = df["siparis_tarihi"].dt.day_name()
df["hafta_numarasi"] = df["siparis_tarihi"].dt.isocalendar().week.astype(int)
df.head()
```

```python
#aylara göre
aylik_yogunluk = (
    df.groupby("siparis_ayi")
    .agg(
        siparis_sayisi = ("siparis_id", "count"),
        benzersiz_musteri_sayisi = ("musteri_id", "nunique")
    )
    .reset_index()
    .sort_values("siparis_ayi")
)
aylik_yogunluk
```

```python
df["ay_donemi"] = df["siparis_gunu"].apply(
    lambda x: "Ayın ilk yarısı" if x <= 15 else "Ayın ikinci yarısı"
)
df.head()
```

```python
ay_donemi_analiz = (
    df.groupby("ay_donemi")
    .agg(
        toplam_tutar = ("toplam_tutar" , "sum")
    )
    .round(1)
    .reset_index()
)
ay_donemi_analiz
```

```python
df["hafta_tipi"] = df["siparis_tarihi"].dt.dayofweek.apply(
    lambda x: "hafta sonu" if x >= 5 else "hafta içi"
)
df.head()
```

```python
hafta_analizi = (
    df.groupby("hafta_tipi")
    .agg(
        ortalama_siparis_tutari = ("toplam_tutar" , "mean"),
    )
    .round(1)
    .reset_index()
)
hafta_analizi
```
